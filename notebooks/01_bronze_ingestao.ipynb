```python
# Célula 1 — Leitura do arquivo bruto
RAW_FILE_PATH = "/Volumes/workspace/default/raw/customer_support_tickets - original.csv"

df_bronze = (spark.read
    .option("header", "true")
    .option("inferSchema", "false")
    .option("multiLine", "true")
    .option("escape", '"')
    .csv(RAW_FILE_PATH)
)

# Renomeando colunas: substitui espaços por underscore
df_bronze = df_bronze.toDF(*[c.replace(" ", "_") for c in df_bronze.columns])

print(f"Total de linhas: {df_bronze.count()}")
print(f"Colunas: {df_bronze.columns}")
```

```python
# Célula 2 — Salvando como Delta na camada Bronze
BRONZE_PATH = "/Volumes/workspace/default/raw/bronze/"

(df_bronze.write
    .format("delta")
    .mode("overwrite")
    .option("overwriteSchema", "true")
    .save(BRONZE_PATH)
)

print("✅ Camada Bronze gravada com sucesso!")
```

```python
# Célula 3 — Validação e visualização
df_check = spark.read.format("delta").load(BRONZE_PATH)
print(f"Linhas na Bronze: {df_check.count()}")
df_check.show(10, truncate=50)
```
