### Célula 1 — Registrando as tabelas Gold no Unity Catalog

```python
import pyspark.sql.functions as F

GOLD_PATH = "/Volumes/workspace/default/raw/gold/"

tabelas = [
    "f_customer_support_tickets",
    "dim_customer",
    "dim_product",
    "dim_type",
    "dim_subject",
    "dim_status",
    "dim_priority",
    "dim_channel",
    "dim_ticket_description",
    "dim_calendario"
]

for tabela in tabelas:
    df = spark.read.format("delta").load(f"{GOLD_PATH}{tabela}/")   # lê o Delta
    df.write \
        .format("delta") \
        .mode("overwrite") \
        .option("overwriteSchema", "true") \
        .saveAsTable(f"default.{tabela}")                            # salva como tabela gerenciada
    
    count = spark.sql(f"SELECT COUNT(*) FROM default.{tabela}").collect()[0][0]
    print(f"✅ default.{tabela} — {count:,} linhas")

print()
print("🎉 Todas as tabelas registradas!")
```
